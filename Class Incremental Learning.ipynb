1. Settings > Accelerator: **GPU P100** (one GPU, fastest fp32 on the free tier).

3. Add Data: attach a CIFAR-100 python dataset (folder `cifar-100-python` with `train`, `test`, `meta`).

Plans: `smoke` (minutes, tests the whole path), `session1`, `session2`, `session3`. See `kaggle_runner.py` for what each runs.

```python
PLAN = "smoke"          # smoke | session1 | session2 | session3
SESSION_HOURS = 11.0    # Kaggle stops committed runs at about 12 h
REPO = "https://github.com/AlbertoEusebio/CIL.git"
BRANCH = "main"
```
